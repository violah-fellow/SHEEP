```python
import duckdb
import pandas as pd

# read_only=True avoids locking the file if pipeline_publications.py happens to be running
db = duckdb.connect('publications.db', read_only=True)

tables = db.sql("SHOW TABLES").df()['name'].tolist()
print(tables)

# schema of every table, for a quick structural overview
for t in tables:
    print(f"\n--- {t} ---")
    display(db.sql(f"DESCRIBE {t}").df())

publications_classified = db.sql("SELECT * FROM publications_classified").df()
publications_classified

# picks up every run table (both original-search RUN_TABLE and REVERSE_TABLE, since both are named run_*)
run_table_names = [t for t in tables if t.startswith('run_')]

run_tables = {}
for t in run_table_names:
    df = db.sql(f"SELECT * FROM {t}").df()
    run_tables[t] = df
    globals()[t] = df  # e.g. creates a variable named run_260703_0726 you can open directly in Data Wrangler

print(f"Loaded {len(run_tables)} run tables: {list(run_tables.keys())}")

db.close()
```
